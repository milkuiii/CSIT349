```python
import pandas as pd

ds = pd.read_csv('Iris.csv')
```

```python
import matplotlib as plt
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

# scaling the data

scaler = StandardScaler()

features_to_scale = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
ds[features_to_scale] = scaler.fit_transform(ds[features_to_scale])

ds.to_csv('Iris_scaled.csv', index=False)

# CHECKING IF THE DATA IS SCALED
print(ds[features_to_scale].describe().round(2))

```

```python
kmeans = KMeans(n_clusters=3, random_state=0)
kmeans.fit(ds[features_to_scale])

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

```
