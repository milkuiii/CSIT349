## Part C -- Names and assignment (slides 12-16)

- Assignment binds a **name** to an **object**. Names have no type; objects do.
- Valid names: letters, digits, underscores; **cannot start with a digit**; case-sensitive (`bob` is not `Bob`).
- Multiple assignment: `x, y = 2, 3` and swap with `x, y = y, x`
- Chained assignment: `a = b = x = 2`
- Using a name before it is assigned raises `NameError`

## Part D -- Sequence types (slides 17-28)

Three sequence types share the same access syntax:

| Type | Example | Mutable? |
|------|---------|----------|
| **tuple** | `("john", 32, ["CMSC"])` | no |
| **string** | `"John Smith"` | no |
| **list** | `[1, 2, "john", ("up", "down")]` | yes |

Items can be mixed types. Indexing is **0-based**. Negative indices count from the right (`-1` is the last item).

**Slicing** `seq[start:stop]` copies from `start` up to **but not including** `stop`.
- `t[:2]` -- from the beginning
- `t[2:]` -- through the end
- `t[:]` -- a copy of the whole sequence

`in` tests membership (`"cd" in "abcde"` is `True`, but `"ac" in "abcde"` is `False`).  
`+` concatenates. `*` repeats.

### Problem 12 -- Copy vs two names for one list

`l2 = l1` makes two names for the **same** list.  
`l3 = l1[:]` makes an **independent copy**.

You will see this again with mutability. For now, just confirm the idea with `id(...)` (the identity of the object in memory).


````python
# Problem 12
l1 = [10, 20, 30]
l2 = l1
l3 = l1[:]

print("l1 is l2?", l1 is l2)   # True: same object
print("l1 is l3?", l1 is l3)   # False: different object
print(id(l1), id(l2), id(l3))

# TODO: predict these two Booleans, then print them
# same_content_copy = (l1 == l3)   # True or False?
# same_object_copy = (l1 is l3)    # True or False?
````
